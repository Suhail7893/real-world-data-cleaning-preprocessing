```python
# Count missing values
missing_values = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (df.isnull().mean() * 100).round(2)

# Create missing value report
missing_report = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": missing_percentage
})

# Display columns containing missing values
display(
    missing_report[missing_report["Missing Count"] > 0]
    .sort_values(by="Missing Percentage", ascending=False)
)
```

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include="object").columns

# Display unique values
for col in categorical_columns:
    print("\nColumn:", col)
    print(df[col].value_counts().head(10))
```

```python
# Count duplicates before cleaning
duplicates_before = df.duplicated().sum()

print("Duplicates Before Cleaning:", duplicates_before)

# Remove duplicate records
df = df.drop_duplicates().copy()

# Count duplicates after cleaning
duplicates_after = df.duplicated().sum()

print("Duplicates After Cleaning:", duplicates_after)

print("Updated Dataset Shape:", df.shape)
```

```python
print("Missing Values Before Treatment:")
display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Fill missing children values using mode
df["children"] = df["children"].fillna(df["children"].mode()[0])

# Replace missing country values
df["country"] = df["country"].fillna("Unknown")

# Replace missing agent values with 0
df["agent"] = df["agent"].fillna(0)

# Drop company column because it has a very high percentage
# of missing values, if company-level analysis is not required
df = df.drop(columns=["company"])

print("Missing value treatment completed!")
```

```python
print("Remaining Missing Values:")

remaining_missing = df.isnull().sum()

display(remaining_missing[remaining_missing > 0])

print("Total Missing Values:", df.isnull().sum().sum())
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

# Remove leading and trailing spaces
for col in categorical_columns:
    df[col] = df[col].str.strip()

print("Categorical data standardization completed!")
```

```python
from sklearn.preprocessing import OneHotEncoder

# Select categorical columns
encoding_columns = [
    "hotel",
    "meal",
    "market_segment",
    "distribution_channel",
    "reserved_room_type",
    "assigned_room_type",
    "deposit_type",
    "customer_type",
    "arrival_date_month"
]

# Initialize encoder
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False,
    dtype=np.uint8
)

# Transform categorical data
encoded_array = encoder.fit_transform(df[encoding_columns])

# Get encoded column names
encoded_column_names = encoder.get_feature_names_out(encoding_columns)

# Create encoded DataFrame
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_column_names,
    index=df.index
)

# Remove original categorical columns
df = df.drop(columns=encoding_columns)

# Combine encoded data with original data
df = pd.concat([df, encoded_df], axis=1)

print("One-Hot Encoding Completed!")
print("New Dataset Shape:", df.shape)

display(df.head())
```
